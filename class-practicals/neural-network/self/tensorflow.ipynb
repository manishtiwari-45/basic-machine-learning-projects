### Forward Propagation + Dense Layer + Vectorization (NumPy + TensorFlow)

```python
import numpy as np
import tensorflow as tf
# for building real neural networks
from tensorflow.keras.models import Sequential 
from tensorflow.keras.layers import Dense
```

```python
#sigmoid function
def sigmoid(z):
    return 1 / (1 + np.exp(-z))
```

```python
# Dense Layer Forward Propagation (Loop Version) 

def dense(a_in, W, b):
    
    units = W.shape[1]          # number of neurons
    a_out = np.zeros(units)     # output array
    
    for j in range(units):
        w = W[:, j]             # take weight column for neuron j
        z = np.dot(w, a_in) + b[j]
        a_out[j] = sigmoid(z)
    
    return a_out
```

```python
a_in = np.array([200, 17])

W = np.array([
    [1, -3, 5],
    [-2, 4, -6]
])

b = np.array([-1, 1, 2])

output = dense(a_in, W, b)

print("Dense Layer Output =", output)
```

```python

# vectorised dense layer

def dense_vectorized(A_in, W, B):
    
    Z = np.matmul(A_in, W) + B
    A_out = sigmoid(Z)
    
    return A_out
```

```python
A_in = np.array([[200, 17]])   # 2D input (1 example, 2 features)

W = np.array([
    [1, -3, 5],
    [-2, 4, -6]
])

B = np.array([[-1, 1, 2]])

print("Vectorized Output =", dense_vectorized(A_in, W, B))
```

```python
# Sequential Forward Propagation (Multiple Layers)

def sequential_forward(x, W1, b1, W2, b2):
    
    a1 = dense_vectorized(x, W1, b1)
    a2 = dense_vectorized(a1, W2, b2)
    
    return a2
```

```python
# TensorFlow Neural Network (Dense Layers)

model = Sequential([
    tf.keras.Input(shape=(2,)),     # 2 input features
    Dense(units=3, activation="sigmoid"),
    Dense(units=1, activation="sigmoid")
])
```

```python
x_train = np.array([
    [200, 17],
    [120, 5],
    [425, 20],
    [212, 18]
])

y_train = np.array([1, 0, 0, 1])
```

```python
model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)
```

```python
model.fit(x_train, y_train, epochs=50)
```

```python
x_new = np.array([[250, 15]])

prediction = model.predict(x_new)

print("Predicted Probability =", prediction)
```
