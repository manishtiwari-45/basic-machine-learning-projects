```python
import numpy as np

x = np.array([2,3])
w = np.array([0.4,0.6])
b = 0.5


z = np.dot(w,x) + b

a = 1/(1+np.exp(-z))

print(z)
print(a)
```

```python
import numpy as np

x = np.array([197,184,136,214])

W1 = np.random.randn(3,4)
b1 = np.random.randn(3,1)

z1 = W1.dot(x) + b1
a1 = 1/(1+np.exp(-z1))

W2 = np.random.randn(1,3)
b2 = np.random.randn(1,1)

z2 = W2.dot(a1) + b2
a2 = 1/(1+np.exp(-z2))

print(a2)
```

```python
import tensorflow as tf
import numpy as np

X = np.array([
[200,17],
[120,5],
[425,20],
[212,18]
])

y = np.array([1,0,0,1])

model = tf.keras.Sequential([
tf.keras.Input(shape=(2,)),
tf.keras.layers.Dense(3, activation='sigmoid'),
tf.keras.layers.Dense(1, activation='sigmoid')
])

model.compile(
optimizer='adam',
loss='binary_crossentropy'
)

model.fit(X,y)

# pred = model.predict([[250,15]])
# print(pred)
```
